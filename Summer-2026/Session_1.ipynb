## Loops

A `for` loop repeats a block of code for every item in a sequence. This is the fundamental tool for processing collections of data.
```python
for variable in iterable:
    do something
```
The **iterable** can be a range of numbers, a list, a string, a dictionary — anything Python can step through one item at a time.

###`range()` — generating a sequence of integers

`range(stop)` produces integers from `0` up to (but **not including**) `stop`.  


````python
# range(stop) — 0 through 4
for i in range(5):
    print(i)
````

`range(start, stop, step)` gives you full control.

````python
# range(start, stop, step)
for i in range(0, 10, 2):   # even numbers
    print(i)
````

You don't always need `range`. When you only need the *values* (not the positions), loop over the list directly — it reads more naturally.

````python
# Working on a list

nums=[3,4,5,6,7] # defining a list

squared_vals=[]

for num in nums:
  squared_vals.append(num**2) # making another list

print(nums)
print(squared_vals)
````

###`enumerate()`— index AND value together

`enumerate()` function in Python is used to iterate over an iterable while keeping track of both the index and the value. Removes the need to manually maintain a counter variable during iteration.

````python
a = ["A", "B", "C"]
r = list(enumerate(a))
print(r)
````

very useful when you need to know the position of each item.This avoids the awkward `range(len(list))` pattern.

````python
# Without enumerate — works but clunky
nums = [10, 20, 30, 40]
for i in range(len(nums)):
    print(f"Index {i}: {nums[i]}")
````

````python
nums = [10, 20, 30, 40]

for index, value in enumerate(nums):
    print(f"Index {index}: {value}")
````

Chemistry example

````python
elements =       ["Lead", "Mercury", "Arsenic", "Cadmium", "Iron"]
concentrations = [0.05,    0.9,       0.3,       0.002,     1.2]   # mg/L

threshold = 0.1  # mg/L — safe limit

for index, element in enumerate(elements):
    if concentrations[index] > threshold:
        print(f"Sample {index}: {element} exceeds threshold at {concentrations[index]} mg/L")
````

###`zip()` — looping over two lists in parallel

`zip()` pairs up elements from two (or more) lists, stopping when the shortest one runs out. This is extremely useful when your data lives in separate parallel lists.

````python
names = ["Alice", "Bob", "Carol"]
scores = [88, 75, 95]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

### Basic `if` statement

````python
import random

temperature = random.randint(25, 35) #Generates an integer between 25 and 35

print(temperature)
if temperature > 30:
    print("It's hot outside.")
````

### `if / elif / else` — multiple branches

````python
score = random.randint(60, 100)

print(score)

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("F")
````

````python
# Categorizing data — combines loop, conditional, and accumulation
grades = [88, 45, 92, 61, 77, 95, 53, 83, 92]

passing = []
failing = []

for g in grades:
    if g >= 60:
        passing.append(g)  #append adds the value to the passing list
    else:
        failing.append(g)

print(f"Passing: {passing}")
print(f"Failing: {failing}")
print(f"Pass rate: {len(passing)/len(grades)*100:.1f}%")
````

Chemistry Example

````python
#Combining for loops, zip(), and conditional statements:
compounds    = ["NaCl",  "H2O",  "HCl",   "NaOH",  "CO2"]
pH_values    = [7.0,      7.0,    1.5,     13.0,     4.5]

results = []

for compound, pH in zip(compounds, pH_values):
    if pH < 7:
        category = "Acidic"
    elif pH > 7:
        category = "Basic"
    else:
        category = "Neutral"

    results.append({
        "compound":    compound,
        "pH":          pH,
        "category":    category
    })

for r in results:
    print(r)
````

### Returning multiple values

````python
import math

def cylinder_stats(radius, height):
    volume = math.pi * radius ** 2 * height
    surface_area = 2 * math.pi * radius * height + 2 * math.pi * radius ** 2
    return volume, surface_area   # Python returns a tuple

vol, area = cylinder_stats(4, 5)
print(f"Volume:       {vol:.2f}")
print(f"Surface area: {area:.2f}")
````

Gibbs free energy example

````python
import math

def gibbs_free_energy(delta_H, delta_S, temp_celsius):
    temp_kelvin = temp_celsius + 273.15                    # convert to Kelvin
    delta_G = delta_H - (temp_kelvin * delta_S)            # ΔG = ΔH - TΔS

    if delta_G < 0:
        spontaneity = "Spontaneous"
    elif delta_G > 0:
        spontaneity = "Non-Spontaneous"
    else:
        spontaneity = "At Equilibrium"

    return temp_kelvin, delta_G, spontaneity


# Thermodynamic data for common reactions
reactions  = ["Combustion of CH4", "Melting of Ice",  "Dissolving NaCl", "Rusting of Iron"]
delta_H    = [-890.3,               6.01,              3.88,              -824.2]   # kJ/mol
delta_S    = [-242.8,               22.0,              43.1,              -270.0]   # J/mol·K
temps      = [25,                   0,                 25,                25]       # °C

print(f"{'Reaction':<22} {'Temp (K)':<12} {'ΔG (kJ/mol)':<15} {'Spontaneity'}") #Titles
print("-" * 65)

for reaction, H, S, temp in zip(reactions, delta_H, delta_S, temps):
    S_kJ = S / 1000      # convert J to kJ to match delta_H units
    temp_K, G, spontaneity = gibbs_free_energy(H, S_kJ, temp)
    print(f"{reaction:<22} {temp_K:<12.2f} {G:<15.2f} {spontaneity}")
````

### f-strings — the modern way to format output

An **f-string** (formatted string literal) lets you embed expressions directly inside a string using curly braces `{}`.

```python
f"text {expression} text"
f"text {expression:.format_spec} text"
```

````python
name = "Bob"
score = 95

# Simple embedding
print(f"Student {name} scored {score}%")

# Expression inside the braces
print(f"Double the score: {score * 2}")

# Calling a function inside
print(f"Name in uppercase: {name.upper()}")
````

#### f-string format specifiers

Format specifiers go after a `:` inside the braces: `{value:spec}`

| Specifier | Meaning | Example | Result |
|-----------|---------|---------|--------|
| `.2f` | 2 decimal places (float) | `{3.14159:.2f}` | `3.14` |
| `.4f` | 4 decimal places | `{3.14159:.4f}` | `3.1416` |
| `d` | integer | `{42:d}` | `42` |
| `05d` | integer, width 5, zero-padded | `{42:05d}` | `00042` |
| `,` | thousands separator | `{1000000:,}` | `1,000,000` |
| `e` | scientific notation | `{12345.6:.2e}` | `1.23e+04` |
| `%` | percentage | `{0.753:.1%}` | `75.3%` |

````python
pi = 3.141592653589793

print(f"Default:        {pi}")
print(f"2 decimals:     {pi:.2f}")
print(f"4 decimals:     {pi:.4f}")
print(f"Scientific:     {pi:.3e}")
````

## File Handling
Basic file handling can be accomplished using the native `open()` function. E.g., `open('name_of_file', 'mode')`. The `'mode'` parameter determines if you will be reading, writing, etc to the file. Here are a few common modes:

| Mode | Meaning |
|------|---------|
| `'r'` | Read |
| `'w'` | Write (overwrites existing content!) |
| `'a'` | Append (adds to existing content) |

The `with` statement automatically closes the file when the block ends, even if an error occurs. **Never use bare `open()` without a matching `close()`.**

````python
# Writing to a file
with open("sample.txt", "w") as f:
    f.write("Line one\n")
    f.write("Line two\n")
    f.write("Line three\n")

print("File written.")
````

````python
# Reading from a file
with open("sample.txt", "r") as f:
    content = f.read()

print(content)
````

````python
# 'a' mode adds to the end without erasing existing content
with open("sample.txt", "a") as f:
    f.write("Line four\n")
    f.write("Line five\n")

# Read it back to confirm
with open("sample.txt", "r") as f:
    print(f.read())
````

# Matplotlib Tutorial

Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python. It's widely used for generating high quality, publication worthy plots for data exploration, analysis, and presentation.

## 1. Importing Matplotlib and Basic Plotting

````python
import matplotlib.pyplot as plt
import numpy as np

# Creating sample data: A sine wave
x = np.linspace(0, 10, 100) # generates 100 evenly spaced values from 0 to 10 on the x axis
y = np.sin(x) # calculates the values of y = sin(x) for each {x} point
````

### Step 1: Establish the 'figure'
Matplotlib calls the whole of the plotting area the 'figure', which may contain
one or more 'axes': areas where data points can be specified and plotted in terms
of their coordinates. E.g., Cartesian x-y coordinates, polar coordinates, etc. The figure can be established with the following command:

````python
plt.figure(figsize=(8, 4)) # Establishes plotting area of 8 x 4 in^2 (width, height)
````

### Step 2: Plot the data
The {x} and {y} data generated earlier with numpy can be plotted using the 'plot' command:

````python
plt.plot(x, y)
````

### Step 3: Add basic labels

````python
plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
````

### Explanation:
*   `import matplotlib.pyplot as plt`: This is the standard way to import the `pyplot` module, which provides a MATLAB-like interface for plotting.
*   `import numpy as np`: We often use NumPy to generate numerical data for plotting.
*   `plt.plot(x, y)`: This function creates a line plot of `y` versus `x`.
*   `plt.title()`, `plt.xlabel()`, `plt.ylabel()`: These functions are used to add a title and labels to the axes.
*   `plt.grid(True)`: Adds a grid to the plot for better readability.
*   `plt.show()`: Displays the plot. Without this, the plot might not appear depending on your environment.

## 2. Customizing Plots: Colors, Markers, and Line Styles

````python
# Generating more data
x2 = np.linspace(0, 10, 30) # generates 30 evenly spaced points from 0 to 10
y2 = np.cos(x2) # calculates y2 = cos({x2})
y3 = np.sin(x2**2) # calculates y3 = sin({x2}^2)

# Establishing a 10 x 6 in^2 figure
plt.figure(figsize=(10, 6))

# Plotting the initial sine wave data with custom color, line style, and marker
plt.plot(x, y, color='red', linestyle='--', label='Sine Wave')

# Adding the cosine wave data to the same figure with a different marker and line style
plt.plot(x2, y2, color='blue', marker='o', linestyle=':', label='Cosine Wave')

# Adding the quadratic sine data to the same figure with a different marker and line syle
plt.plot(x2, y3, color='green', marker='s', linestyle='-.', label='Quadratic Sine')

plt.title('Customized Plots')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend() # Display the legend
plt.grid(True)
plt.show()
````

### Customization Parameters: arguments of `plt.plot()`
*   `color`: Sets the color of the line/markers (e.g., `'red'`, `'blue'`, `'#FF4500'`).
*   `linestyle` (or `ls`): Sets the line style (e.g., `'-'` for solid, `'--'` for dashed, `':'` for dotted, `'-.'` for dash-dot).
*   `marker`: Sets the marker style for data points (e.g., `'o'` for circle, `'x'` for x, `'^'` for triangle, `'s'` for square).
*   `label`: A string that will be used in the legend.

## 3. Creating a Figure with Multiple Subplots

````python
# Generating data using numpy
x = np.linspace(0, 5, 100) # 100 evenly spaced points from 0 to 5
y1 = x**2
y2 = np.exp(x)

# Establishing a figure with two subplots side by side, overall area of 12 x 5 in^2
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Plot on the first axes (left)
ax[0].plot(x, y1, color='purple')
ax[0].set_title(r'$Quadratic$' + ' ' + r'$Function$')
ax[0].set_xlabel(r'$X$')
ax[0].set_ylabel(r'$Y = X^2$')
ax[0].grid(True)

# Plot on the second axes (right)
ax[1].plot(x, y2, color='darkgreen')
ax[1].set_title(r'$Exponential$' + ' ' + r'$Function$')
ax[1].set_xlabel(r'$X$')
ax[1].set_ylabel(r'$Y = e^{X}$')
ax[1].grid(True)

plt.tight_layout() # Adjusts subplot params for a tight layout
plt.show()
````

### Explanation:
*   `fig, ax = plt.subplots(1, 2, figsize=(12, 5))`: This creates a figure (`fig`) and a set of subplots, aka axes (`ax`). Here, `1, 2` means 1 row and 2 columns of subplots. `figsize` sets the overall size of the figure.
*   A particular axes, `ax`, is a NumPy array of axes objects. We can access individual subplots using `ax[0]`, `ax[1]`, etc.
*   Instead of `plt.plot()`, `plt.title()`, etc., we use the methods of the `ax` objects (e.g., `ax[0].plot()`, `ax[0].set_title()`).
*   `plt.tight_layout()`: Automatically adjusts subplot parameters for a tight layout, preventing labels from overlapping.

### Note:


# NumPy in practice:  data-analysis workflow

We'll take one messy data
file (with missing values and a bad measurement, like real data) and walk the whole pipeline
a scientist runs every day:


*   load
*   inspect
*   clean
*   describe
*   filter
*   normalize
*   find relationships  
*   aggregate
*   smooth




````python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(7)
````

### Let's create an example data file

We'll generate a small CSV so the notebook is self-contained — pretend it just came off your
instrument. It has 60 experimental runs and four columns:
`temperature, catalyst_conc, yield, purity`.

````python
n = 60
temperature   = rng.uniform(300, 500, n)        # Kelvin
catalyst_conc = rng.uniform(1, 10, n)           # mM
yield_pct = 100*(1 - np.exp(-(temperature-300)/90)) + 1.5*catalyst_conc + rng.normal(0,4,n)
yield_pct = np.clip(yield_pct, 0, 100)
purity    = 99 - 0.03*(temperature-300) + rng.normal(0, 1.0, n)

# inject realistic problems:
yield_pct[[7, 23, 41]] = np.nan      # instrument dropouts -> missing values
yield_pct[15] = 250.0                # a mis-recorded outlier (impossible yield)
purity[31]    = 60.0                 # a bad measurement

with open("experiments.csv", "w") as f:
    f.write("temperature,catalyst_conc,yield,purity\n")
    for t, c, y, p in zip(temperature, catalyst_conc, yield_pct, purity):
        ys = "" if np.isnan(y) else f"{y:.2f}"     # blank cell for missing values
        f.write(f"{t:.2f},{c:.2f},{ys},{p:.2f}\n")

print("wrote experiments.csv — first few lines:")
print("".join(open("experiments.csv").readlines()[:5]))
````

### 1. Load a data file

`np.genfromtxt` reads a text/CSV file into an array. Blank cells automatically become `nan`
("not a number") — exactly what we want for missing data.

````python
data = np.genfromtxt("experiments.csv", delimiter=",", skip_header=1)
print("shape:", data.shape)          # (60 rows, 4 columns)
print("first 3 rows:\n", data[:3])

# Give the columns readable names (column slicing):
temperature   = data[:, 0]
catalyst_conc = data[:, 1]
yield_pct     = data[:, 2]
purity        = data[:, 3]
````

### 2. Describe your data

Summary statistics in one line each. **Watch out for the missing values:**

````python
print("plain mean of yield :", yield_pct.mean())      # nan! one missing value poisons it
print("nan-aware mean      :", np.nanmean(yield_pct)) # ignores the missing entries

print()
print(f"temperature: mean {temperature.mean():.0f} K, "
      f"range {temperature.min():.0f}–{temperature.max():.0f} K")
print(f"purity:      median {np.median(purity):.1f}%, std {purity.std():.2f}")
print("yield quartiles (25/50/75%):", np.round(np.nanpercentile(yield_pct, [25,50,75]), 1))
````

ordinary `.mean()` returns `nan` if *any*
value is missing. Use the `nan`-aware versions (`np.nanmean`, `np.nanstd`, `np.nanpercentile`)
on real data.

````python
### 3. Clean it: missing values & outliers

First, count the missing values. Then flag outliers with the **IQR rule** (anything far
outside the middle 50% of the data) — a robust, standard method.
````

````python
missing = np.isnan(yield_pct)
print("missing yield values:", missing.sum())

# IQR outlier rule on the values we DO have
y = yield_pct[~missing]
q1, q3 = np.percentile(y, [25, 75])
iqr = q3 - q1
low, high = q1 - 1.5*iqr, q3 + 1.5*iqr
outliers = (yield_pct < low) | (yield_pct > high)
print("outliers flagged:", outliers.sum(), "-> values:", yield_pct[outliers])

# Make a cleaned copy: turn the outlier into 'missing' so stats ignore it
clean_yield = yield_pct.copy()
clean_yield[outliers] = np.nan
print("clean mean yield:", round(np.nanmean(clean_yield), 1), "%")
````

the impossible 250% yield gets caught and removed without you eyeballing
60 numbers. The IQR rule scales to millions of rows. `.copy()` matters — edit a copy so you
never lose the raw data.

### 4. Filter & subset rows

"Show me only the runs that meet these conditions" — the most common data question there is.
Combine conditions with `&` (and) / `|` (or), and **wrap each condition in parentheses**.

````python
# high-yield runs at high temperature
mask = (clean_yield > 80) & (temperature > 420)
print("runs matching (yield>80 AND T>420):", mask.sum())
print("their average catalyst conc:", round(catalyst_conc[mask].mean(), 2), "mM")

# conditional summary: mean purity, low vs high temperature
hot = temperature > 400
print(f"mean purity  hot runs: {purity[hot].mean():.1f}% | "
      f"cool runs: {purity[~hot].mean():.1f}%")
````

### 5. Normalize & standardize

Different variables live on different scales (Kelvin vs mM vs %). Before comparing them — or
feeding them to a machine-learning model — you put them on a common footing.

````python
# z-score: mean 0, std 1 (the standard ML preprocessing)
temp_z = (temperature - temperature.mean()) / temperature.std()

# min-max: squash to the range [0, 1]
temp_01 = (temperature - temperature.min()) / (temperature.max() - temperature.min())

print(f"original T : {temperature[:4].round(0)}")
print(f"z-scored   : {temp_z[:4].round(2)}   (mean {temp_z.mean():.1e}, std {temp_z.std():.2f})")
print(f"min-max    : {temp_01[:4].round(2)}   (min {temp_01.min():.0f}, max {temp_01.max():.0f})")
````

### 6. Find relationships: correlation & trend line

Does yield actually depend on temperature? Measure it (`np.corrcoef`) and draw the trend
(`np.polyfit`).

````python
good = ~np.isnan(clean_yield)                       # use only valid rows
r = np.corrcoef(temperature[good], clean_yield[good])[0, 1]
print(f"correlation(temperature, yield) = {r:.2f}")

slope, intercept = np.polyfit(temperature[good], clean_yield[good], 1)
print(f"trend: yield ≈ {slope:.2f} × T + {intercept:.1f}")

plt.figure(figsize=(7, 4))
plt.scatter(temperature[good], clean_yield[good], s=25, label="runs")
xs = np.array([temperature.min(), temperature.max()])
plt.plot(xs, slope*xs + intercept, color="red", label="trend")
plt.xlabel("temperature (K)"); plt.ylabel("yield (%)")
plt.title(f"yield vs temperature  (r = {r:.2f})")
plt.legend(); plt.show()
````

`r` near +1 means "go up together", near 0 means "no linear link", near −1
means "one up, the other down". `np.corrcoef` returns the full correlation matrix; we grabbed
the off-diagonal `[0,1]` entry.

### 7. Aggregate: group values into bins

"Average yield per temperature range" — a group-by, done with `np.digitize` to label each row
with its bin, then a mask per group.

 a mask per group plus a reduction:

````python
edges = np.array([300, 350, 400, 450, 500])
which_bin = np.digitize(temperature, edges)        # which interval each run falls into

print("temperature range   mean yield   n")
for b in range(1, len(edges)):
    sel = (which_bin == b) & ~np.isnan(clean_yield)
    print(f"  {edges[b-1]}–{edges[b]} K        "
          f"{clean_yield[sel].mean():5.1f}%     {sel.sum()}")
````

**pandas `groupby`** is more convenient for this:

````python
import pandas as pd

rng = np.random.default_rng(7)
n = 60
temperature   = rng.uniform(300, 500, n)
catalyst_conc = rng.uniform(1, 10, n)
yield_pct = 100*(1 - np.exp(-(temperature-300)/90)) + 1.5*catalyst_conc + rng.normal(0,4,n)
yield_pct = np.clip(yield_pct, 0, 100)
purity = 99 - 0.03*(temperature-300) + rng.normal(0,1.0,n)
yield_pct[[7,23,41]] = np.nan; yield_pct[15] = 250.0; purity[31] = 60.0
with open("experiments.csv","w") as f:
    f.write("temperature,catalyst_conc,yield,purity\n")
    for t,c,y,p in zip(temperature,catalyst_conc,yield_pct,purity):
        ys = "" if np.isnan(y) else f"{y:.2f}"
        f.write(f"{t:.2f},{c:.2f},{ys},{p:.2f}\n")
print("data ready")
````

### 8. Smooth a noisy signal

Instrument traces, sensor logs, and simulation observables are noisy. A **moving average**
(via `np.convolve`) reveals the underlying trend.

````python
t = np.linspace(0, 20, 500)
signal = np.sin(t) + 0.3*t + rng.normal(0, 0.4, t.size)   # oscillation + drift + noise

k = 21                                  # window size
kernel = np.ones(k) / k                 # equal weights that sum to 1
smooth = np.convolve(signal, kernel, mode="same")

plt.figure(figsize=(9, 4))
plt.plot(t, signal, lw=0.8, alpha=0.5, label="raw signal")
plt.plot(t, smooth, lw=2.2, label=f"{k}-point moving average")
plt.xlabel("Time"); plt.ylabel("measurement")
plt.title("Smoothing a noisy trace")
plt.legend(); plt.show()
````

smoothing before peak-finding, baseline estimation, or just to see a trend.
The window size is a knob: bigger = smoother but more lag. (Edges are slightly off because the
window runs out of data there — a known trade-off.)
 *Try:* change `k` to `5` (barely smoothed) or `61` (very smooth, sluggish).
